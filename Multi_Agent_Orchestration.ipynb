```python

from datetime import datetime, date

# AGENT 1: Meeting Planner
def meeting_planner(request):
    print("\nAgent 1: Meeting Planner is working...")

    if not request["topic"] or not request["attendees"]:
        return {"status": "error", "message": "Meeting details are missing."}

    return {
        "topic": request["topic"],
        "date": request["date"],
        "time": request["time"],
        "attendees": request["attendees"]
    }


# AGENT 2: Time Checker
def time_checker(meeting):
    print("Agent 2: Time Checker is working...")

    try:
        meeting_date = date.fromisoformat(meeting["date"])
        meeting_time = datetime.strptime(
            meeting["time"], "%I:%M %p"
        ).time()

        if meeting_date < date.today():
            return {"status": "error", "message": "Date is in the past."}

        if not (9 <= meeting_time.hour < 18):
            return {
                "status": "warning",
                "message": "Meeting is outside working hours (9 AM–6 PM)."
            }

        return {"status": "success",
                "message": "Date and working hours are valid."}

    except ValueError:
        return {
            "status": "error",
            "message": "Invalid date or time format."
        }


# AGENT 3: Summary Generator
def summary_generator(meeting, validation):
    print("Agent 3: Summary Generator is working...")

    return f"""
===== FINAL MEETING PLAN =====
Topic       : {meeting['topic']}
Date        : {meeting['date']}
Time        : {meeting['time']}
Attendees   : {meeting['attendees']}
Validation  : {validation['message']}
==============================
"""


# ORCHESTRATOR: Coordinates all 3 agents
def orchestrator(request):
    print("Orchestrator started.")

    meeting = meeting_planner(request)

    if meeting.get("status") == "error":
        print(meeting["message"])
        return

    validation = time_checker(meeting)

    if validation["status"] == "error":
        print(validation["message"])
        return

    result = summary_generator(meeting, validation)
    print(result)
    print("Orchestration completed!")


# INPUT: Example use case
request = {
    "topic": "EMEA AI Training Team Meeting",
    "date": "2026-10-16",
    "time": "10:00 AM",
    "attendees": "AI Training Team"
}

# Start the multi-agent workflow
orchestrator(request)
```
